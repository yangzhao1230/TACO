```python
import pandas as pd

# Cells and levels configuration
cells = ["hepg2", "k562", 'sknsh']
levels = ["mbo"]
tfbs_lambdas = [0.0, 0.01, 0.1]
selected_round = 100

# Iterate over all combinations of cells, levels, and tfbs_lambdas
for cell in cells:
    for level in levels:
        for tfbs_lambda in tfbs_lambdas:
            # Construct the file path
            # path = f"summary/random_{tfbs_lambda}_{cell}_{level}.csv"
            path = f"summary/mbo_{tfbs_lambda}_{cell}_{level}.csv"
            try:
                # Read the CSV file
                df = pd.read_csv(path)
                
                # Filter data for the selected round
                last_round_data = df[df['round'] == selected_round]
                
                # Verify the last round is 100
                assert df['round'].max() == 100, "The last round is not 100"
                
                # Compute mean and std for metrics, rounded to two decimals
                mean_values = last_round_data.mean().round(2)
                std_values = last_round_data.std().round(2)
                
                # Remove the first two columns (likely 'round' and 'novelty')
                mean_values = mean_values.iloc[2:]
                std_values = std_values.iloc[2:]
                
                # Print results in LaTeX-style format
                print(f"Results for Cell: {cell}, Level: {level}, TFBS Lambda: {tfbs_lambda}")
                print("Metric (Mean ± Std):")
                for metric, mean in mean_values.items():
                    std = std_values[metric]
                    print(f"{metric}: {mean} ± {std}")
                print("\n" + "-"*50 + "\n")
            
            except FileNotFoundError:
                print(f"File not found for Cell: {cell}, Level: {level}, TFBS Lambda: {tfbs_lambda}")
            except Exception as e:
                print(f"An error occurred for Cell: {cell}, Level: {level}, TFBS Lambda: {tfbs_lambda}: {e}")
```
