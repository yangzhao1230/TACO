```python
import pandas as pd

# cells = ['complex', 'defined']
cells = ["hepg2", "k562", "sknsh"]
levels = ["mbo"]  # 替换为你的 level 列表

selected_round = 100
# 遍历每个组合
for cell in cells:
    for level in levels:
        # 构建路径
        path = f'summary/{cell}_{level}_0.1_total.csv'
        
        try:
            # 读取 CSV 文件
            df = pd.read_csv(path)
            
            # 筛选出最后一个 round 的数据
            # last_round_data = df[df['round'] == df['round'].max()]
            last_round_data = df[df['round'] == selected_round]
            # 验证最后一个 round 是否为 99
            assert df['round'].max() == 100, "The last round is not 100"
            
            # 计算每个 metric 的平均值和标准差，并保留两位小数
            mean_values = last_round_data.mean().round(2)
            std_values = last_round_data.std().round(2)
            
            # 删除前两列数据 (可能是 'round' 和 'novelty' 等不需要的数据)
            mean_values = mean_values.iloc[2:]
            std_values = std_values.iloc[2:]
            
            # 打印当前组合及其结果，格式化为 LaTeX 表格中需要的形式
            print(f"Results for Cell: {cell}, Level: {level}")
            print("Metric (Mean ± Std):")
            for metric, mean in mean_values.items():
                std = std_values[metric]
                print(f"{metric}: {mean} ± {std}")
            print("\n" + "-"*50 + "\n")
            
        except FileNotFoundError:
            print(f"File not found for Cell: {cell}, Level: {level}")
        except Exception as e:
            print(f"An error occurred for Cell: {cell}, Level: {level}: {e}")
```
